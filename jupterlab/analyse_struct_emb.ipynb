```python
OUT_DIR = "/home/gavrilev/DLResearchPPI/PPI Research/src_/embeddings/ppi struct/"
dataset = "dataset1"  # replace with actual dataset name
output_path = os.path.join(OUT_DIR, f"{dataset}_processed_matrix_data.h5")
```

```python
with h5py.File(output_path, "r") as hf:

    # ── 1. Top-level overview ────────────────────────────────────────────────
    print("=== Top-level keys (strategies) ===")
    print(list(hf.keys()))
    # e.g. ['conditional', 'full_mean', 'subset_mean']

    # ── 2. Models available under one strategy ───────────────────────────────
    strategy = "conditional"
    print(f"\n=== Models under '{strategy}' ===")
    print(list(hf[strategy].keys()))
    # e.g. ['model_1', 'model_2']

    # ── 3. PPI IDs available under one model ────────────────────────────────
    model = list(hf[strategy].keys())[0]   # grab first model dynamically
    print(f"\n=== First 5 PPI IDs under '{strategy}/{model}' ===")
    ppi_ids = list(hf[strategy][model].keys())
    print(ppi_ids[:5])

    # ── 4. Features stored for a single PPI ─────────────────────────────────
    ppi_id = ppi_ids[0]
    print(f"\n=== Features stored for '{strategy}/{model}/{ppi_id}' ===")
    print(list(hf[strategy][model][ppi_id].keys()))
    # e.g. ['mean_ifeature_omega_a', 'mean_ifeature_omega_b',
    #        'mean_ppi_former_a', 'mean_ppi_former_b', 'progres_vector']

    # ── 5. Read a single feature array ──────────────────────────────────────
    mean_a = hf[strategy][model][ppi_id]["mean_ppi_former_a"][:]
    print(f"\n=== mean_ppi_former_a for '{ppi_id}' ===")
    print(f"  shape : {mean_a.shape}")
    print(f"  dtype : {mean_a.dtype}")
    print(f"  values: {mean_a}")

    # ── 6. Compare same PPI across all three strategies ──────────────────────
    print(f"\n=== mean_ppi_former_a across strategies for '{ppi_id}' ===")
    for strat in hf.keys():
        arr = hf[strat][model][ppi_id]["mean_ppi_former_a"][:]
        print(f"  [{strat}]  shape={arr.shape}  mean={arr.mean():.6f}  std={arr.std():.6f}")

    # ── 7. Verify progres_vector is hard-linked (same data, no duplication) ──
    print(f"\n=== progres_vector hard-link verification for '{ppi_id}' ===")
    pv_conditional = hf["conditional"][model][ppi_id]["progres_vector"][:]
    pv_full_mean   = hf["full_mean"][model][ppi_id]["progres_vector"][:]
    pv_subset_mean = hf["subset_mean"][model][ppi_id]["progres_vector"][:]
    print(f"  conditional vs full_mean  identical: {np.array_equal(pv_conditional, pv_full_mean)}")
    print(f"  conditional vs subset_mean identical: {np.array_equal(pv_conditional, pv_subset_mean)}")

    # ── 8. Iterate over all PPIs for one strategy/model ─────────────────────
    print(f"\n=== Iterating all PPIs under '{strategy}/{model}' ===")
    for pid in list(hf[strategy][model].keys())[:3]:   # first 3 as a sample
        grp = hf[strategy][model][pid]
        pf_a = grp["mean_ppi_former_a"][:]
        pf_b = grp["mean_ppi_former_b"][:]
        prog = grp["progres_vector"][:]
        print(f"  ppi_id={pid:20s}  pf_a.shape={pf_a.shape}  pf_b.shape={pf_b.shape}  progres.shape={prog.shape}")

    # ── 9. Load everything into memory as a nested dict (mirrors old pickle) ──
    print(f"\n=== Loading full strategy into memory dict ===")
    def h5_to_dict(h5file, strategy):
        result = {}
        for model_key in h5file[strategy].keys():
            result[model_key] = {}
            for ppi_key in h5file[strategy][model_key].keys():
                result[model_key][ppi_key] = {
                    feat: h5file[strategy][model_key][ppi_key][feat][:]
                    for feat in h5file[strategy][model_key][ppi_key].keys()
                }
        return result

    data = h5_to_dict(hf, "conditional")
    print(f"  models loaded : {list(data.keys())}")
    print(f"  PPIs in first model: {len(data[list(data.keys())[0]])}")
```
